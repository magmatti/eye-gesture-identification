### Identification output

Report generation for eye gesture event detection

```python
import pandas as pd
from IPython.display import Markdown, display

from src.analyze import build_report_tables, run_analysis
from src.detection_config import DetectionConfig
from src.plots import plot_saccade_directions
```

### Helper functions

```python
def display_all_events_by_scenario(events: pd.DataFrame) -> None:
    for scenario, group in events.groupby("scenario", sort=True):
        display(Markdown((f"# {scenario}")))
        display(group)
```

### Running analysis

```python
VERBOSE = False

result = run_analysis(DetectionConfig())
report_tables = build_report_tables(result=result, verbose=VERBOSE)
```

### Show data

```python
for label, table in report_tables:
    display(Markdown(f"### {label}"))
    if label == "All events":
        display_all_events_by_scenario(table)
    else:
        display(table)
```
