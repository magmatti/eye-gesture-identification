## Table 6.6

```python
ev = build_blink_evaluation(result.blink_matches, result.events)
ev = ev[(ev["participant"] != "TOTAL") & (ev["scenario"] == "blink")]

t6 = ev.set_index("participant")[["expected_count", "tp", "fp", "fn"]]
t6.loc["RAZEM"] = t6.sum()
t6["recall"] = t6.tp / t6.expected_count
t6["precision"] = t6.tp / (t6.tp + t6.fp)
t6["f1"] = 2 * t6.recall * t6.precision / (t6.recall + t6.precision)

display(t6.round(3))
```
