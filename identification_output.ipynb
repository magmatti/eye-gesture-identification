### Imports

```python
import re

import pandas as pd
from IPython.display import Markdown, display

from src.analyze import build_report_tables, run_analysis
from src.detection_config import DetectionConfig
from src.plots import plot_saccade_directions
```

### Helper functions

```python
def file_type(source_file: str) -> str:
    return re.sub(r"_\d{8}_\d{6}\.csv$", "", str(source_file))


def display_all_events_by_file_type(events: pd.DataFrame) -> None:
    events = events.copy()
    events["file_type"] = events["source_file"].map(file_type)

    for group_name, group in events.groupby("file_type", sort=True):
        display(Markdown(f"#### {group_name}"))
        display(group.drop(columns="file_type"))
```

### Running analysis

```python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

events = run_analysis(DetectionConfig())

report_tables = build_report_tables(events)
```

### Show data

```python
for label, table in report_tables:
    display(Markdown(f"### {label}"))
    if label == "All events":
        display_all_events_by_file_type(table)
    else:
        display(table)
```
