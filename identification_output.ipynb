### Imports

```python
import re

import pandas as pd
from IPython.display import Markdown, display

from src.analyze import run_analysis
from src.detection_config import DetectionConfig
from src.event_detection import (
    summarize_event_counts_by_file,
    summarize_events_by_file,
    summarize_events_by_gesture,
)
```

### Helper functions

```python
def file_type(source_file: str) -> str:
    return re.sub(r"_\d{8}_\d{6}\.csv$", "", str(source_file))


def display_all_events_by_file_type(events: pd.DataFrame) -> None:
    events = events.copy()
    events["file_type"] = events["source_file"].map(file_type)

    for group_name, group in events.groupby("file_type", sort=True):
        display(Markdown(f"#### {group_name}"))
        display(group.drop(columns="file_type"))
```

### Running analysis

```python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

events = run_analysis(DetectionConfig())

report_tables = [
    ("Event counts by file", summarize_event_counts_by_file(events)),
    ("Event summary by file", summarize_events_by_file(events)),
    ("Event summary by gesture", summarize_events_by_gesture(events)),
    ("All events", events),
]
```

### Show data

```python
for label, table in report_tables:
    display(Markdown(f"### {label}"))
    if label == "All events":
        display_all_events_by_file_type(table)
    else:
        display(table)
```
