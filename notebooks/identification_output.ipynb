### Imports

```python
from pathlib import Path
import os
import re

import pandas as pd
from IPython.display import Markdown, display

from src.analyze import REPORTS_DIR, run_analysis
from src.detection_config import DetectionConfig
```

### Helper functions

```python
def find_project_root(start: Path) -> Path:
    for path in (start, *start.parents):
        if (path / "src" / "analyze.py").exists() and (path / "data").exists():
            return path
    raise FileNotFoundError("Could not find the project root containing src/analyze.py and data/.")


def file_type(source_file: str) -> str:
    return re.sub(r"_\d{8}_\d{6}\.csv$", "", str(source_file))


def display_all_events_by_file_type(events: pd.DataFrame) -> None:
    events = events.copy()
    events["file_type"] = events["source_file"].map(file_type)

    for group_name, group in events.groupby("file_type", sort=True):
        display(Markdown(f"#### {group_name}"))
        display(group.drop(columns="file_type"))
```

### Running analysis and saving to reports

```python
project_root = find_project_root(Path.cwd().resolve())
os.chdir(project_root)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

run_analysis(DetectionConfig())

events_dir = REPORTS_DIR / "events"
report_tables = [
    ("Event counts by file", "event_counts_by_file.csv"),
    ("Event summary by file", "event_summary_by_file.csv"),
    ("Event summary by gesture", "event_summary_by_gesture.csv"),
    ("All events", "all_events.csv"),
]
```

### Show data

```python
for label, filename in report_tables:
    display(Markdown(f"### {label}"))
    report = pd.read_csv(events_dir / filename)
    if filename == "all_events.csv":
        display_all_events_by_file_type(report)
    else:
        display(report)
```
